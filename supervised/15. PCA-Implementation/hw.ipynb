```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.decomposition import PCA

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

import tensorflow as tf
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# Flatten (sklearn modellar uchun)
X_train_flat = X_train.reshape(len(X_train), -1)
X_test_flat = X_test.reshape(len(X_test), -1)
```

```python
# 📌 3. Foydali funksiyalar
def evaluate_model(model, X_train, y_train, X_test, y_test, model_name="Model"):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(f"{model_name} Accuracy: {acc:.4f}")
    
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(8,6))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"{model_name} - Confusion Matrix")
    plt.show()
    
    print(classification_report(y_test, y_pred))
    return acc
```

```python
# 📌 4. Oddiy ML modellari (baseline)
acc_logreg = evaluate_model(LogisticRegression(max_iter=1000), X_train_flat, y_train, X_test_flat, y_test, "Logistic Regression")
acc_svm = evaluate_model(SVC(), X_train_flat, y_train, X_test_flat, y_test, "SVM")
acc_knn = evaluate_model(KNeighborsClassifier(), X_train_flat, y_train, X_test_flat, y_test, "KNN")
acc_tree = evaluate_model(DecisionTreeClassifier(), X_train_flat, y_train, X_test_flat, y_test, "Decision Tree")
acc_rf = evaluate_model(RandomForestClassifier(), X_train_flat, y_train, X_test_flat, y_test, "Random Forest")
```

```python
# 📌 5. Ensemble metodlar
acc_ada = evaluate_model(AdaBoostClassifier(), X_train_flat, y_train, X_test_flat, y_test, "AdaBoost")
acc_gb = evaluate_model(GradientBoostingClassifier(), X_train_flat, y_train, X_test_flat, y_test, "GradientBoosting")
acc_xgb = evaluate_model(XGBClassifier(use_label_encoder=False, eval_metric='mlogloss'), X_train_flat, y_train, X_test_flat, y_test, "XGBoost")
acc_lgbm = evaluate_model(LGBMClassifier(), X_train_flat, y_train, X_test_flat, y_test, "LightGBM")
acc_cat = evaluate_model(CatBoostClassifier(verbose=0), X_train_flat, y_train, X_test_flat, y_test, "CatBoost")
```

```python
# 📌 6. PCA bilan dimensionality reduction (100 komponenta)
pca = PCA(n_components=100)
X_train_pca = pca.fit_transform(X_train_flat)
X_test_pca = pca.transform(X_test_flat)
```

```python
acc_svm_pca = evaluate_model(SVC(), X_train_pca, y_train, X_test_pca, y_test, "SVM + PCA")
```
